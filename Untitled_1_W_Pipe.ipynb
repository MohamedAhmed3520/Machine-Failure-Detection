## first we are going to use RandomUnderSampler

```python
#Resample the data
x_res , y_res = RandomUnderSampler().fit_resample(x_train, y_train)
```

```python
y_res.value_counts(normalize=True)
```

```python
# Split the data into training and testing sets
x_tr , x_val , y_tr , y_val = train_test_split(x_res, y_res, test_size=0.2, random_state=42)
```

```python
# Select numerical and categorical columns
num_cols = x.select_dtypes(include=np.number).columns

cat_cols = x.select_dtypes(include=['object', 'category']).columns
```

```python
# Create Numerical  pipeline
num_pipe = Pipeline([
    ('imputer', SimpleImputer(strategy='mean')),
    ('scaler', StandardScaler())
])
```

```python
# Create Categorical pipeline
cat_pipe = Pipeline([
    ('imputer', SimpleImputer(strategy='most_frequent')),
    ('onehot', OneHotEncoder(handle_unknown='ignore'))
])
```

```python
# Combine the pipelines
preprocessor = ColumnTransformer([
    ('num', num_pipe, num_cols),
    ('cat', cat_pipe, cat_cols)
])
```

```python
#Scale the data
x_tr_sc = preprocessor.fit_transform(x_tr)
x_val_sc = preprocessor.transform(x_val)
```

```python
# Fit the model
rfc = RandomForestClassifier()

rfc.fit(x_tr_sc, y_tr)
```

```python
rfc.score(x_val_sc, y_val)
```

```python
print(classification_report(y_val, rfc.predict(x_val_sc)))
```

```python
print(confusion_matrix(y_val, rfc.predict(x_val_sc)))
print(recall_score(y_val, rfc.predict(x_val_sc)))
print(precision_score(y_val, rfc.predict(x_val_sc)))
```
